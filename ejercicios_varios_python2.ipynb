## Calcular Promedio // pedir nota1, nota2 y nota3 . sumar las dos primeras notas y la tercera dividirla entre /3

```python
numero1 = float (input ("Digite la primera nota: "))
numero2 = float (input ("Digite la segunda nota: "))
numero3 = float (input ("Digite la tercera nota: "))
promedio = (numero1+numero2+numero3)/3
print ("El promedio de nota es: ", promedio)
```

## Promedio con desición. pedir nota y si es > a 70 aprobó y si es < reprobó:

```python
numero1 = float (input ("Digite su primer nota: "))
numero2 = float (input ("Digite su segunda nota: "))
promedio = (numero1+numero2)/2
if promedio >= 70: print ("Aprobó")
else: print ("Reprobó")
```
