#sklearn es una biblioteca de python especializada en Machine Learning

```python
from sklearn.tree import DecisionTreeClassifier # aprendizaje supervisado

#Datos - ingreso de los datos - ingreso a una tienda a partir de la posición 4 en rango de a 20-18 años,
#que tan probable es que haga una compra
## Cada fila representa un cliente:
# [edad, número de compras]

X = [
    [18, 1],
    [20, 2],
    [25, 5],
    [30, 7],
    [35, 9]
]

# Posibles respuestas

y = [
    "No",
    "No",
    "Si",
    "Si",
    "Si"
]


#Crear el modelo / # Crear el modelo de árbol de decisión
## fit() significa: "Aprende con estos datos"

modelo = DecisionTreeClassifier ()

# Entrenar el modelo con fit () significa que aprende con los datos brindados
modelo.fit(X,y)

#Nuevo Cliente: se crea el nuevo cliente/ # Tiene 27 años y ha realizado 6 compras
nuevo_cliente = [[27, 6]]

# Hará una Predicción usando modelo.predict. le pedimos a python que nos dé o haga una preddicción
## predict() le pide al modelo que prediga la respuesta
prediccion = modelo.predict (nuevo_cliente)

# Porque predicción [0]? Python empieza a contar desde 0
print ("Esta es la Predicción:" , prediccion [0])





```

```python
# Este comando trae una herramienta que sirve para dividir nuestros datos en dos grupos
from sklearn.model_selection import train_test_split
# Con este comando se importa el árbol de desición
from sklearn.tree import DecisionTreeClassifier
# Esto sirve para medir la exactitud del modelo ¿Cuantas veces acertó la computadora?
from sklearn.metrics import accuracy_score

X = [
    [18, 1],
    [20, 2],
    [22, 3],
    [25, 5],
    [27, 6],
    [30, 7],
    [32, 8],
    [35, 9],
    [40, 10],
    [45, 12]
]

y = [
    "No",
    "No",
    "No",
    "Sí",
    "Sí",
    "Sí",
    "Sí",
    "Sí",
    "Sí",
    "Sí"
]

# Separar datos
#test_size=0.3 Significa: 30% de los datos serán para probar.
#random_state=42 sirve para controlar el componente aleatorio de la división.
#train_test_split mezcla los datos antes de separarlos.
#X_entrenamiento Datos que el modelo utilizará para aprender.
#y_entrenamiento Las respuestas correspondientes a esos datos.
#X_prueba Datos que el modelo no debe utilizar para aprender.
#y_prueba Las respuestas verdaderas que utilizaremos para comprobar si acertó.

X_entrenamiento, X_prueba, y_entrenamiento, y_prueba = train_test_split(
    X, y, test_size=0.3, random_state=42
)

#Crear el modelo / # Crear el modelo de árbol de decisión
## fit() significa: "Aprende con estos datos"

modelo = DecisionTreeClassifier ()

# Entrenar el modelo con fit () significa que aprende con los datos brindados
modelo.fit(X_entrenamiento,y_entrenamiento)

# Hará una Predicción usando modelo.predict. le pedimos a python que nos dé o haga una preddicción
## predict() le pide al modelo que prediga la respuesta
prediccion = modelo.predict (X_prueba)

# Medir la exactitud
precision = accuracy_score (y_prueba, prediccion)

print("Predicción del modelo:",  prediccion)
print("Respuestas Reales:",  y_prueba)
print("Precisión del modelo:",  precision)
```
