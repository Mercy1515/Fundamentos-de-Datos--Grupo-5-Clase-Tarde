```python
import pandas as pd                       # Librería para manejar datos en tablas (DataFrames)
import numpy as np                        # Librería para cálculos numéricos y arreglos
import matplotlib.pyplot as plt           # Librería para crear gráficos
from sklearn.linear_model import LinearRegression  # Modelo de regresión lineal
from google.colab import files            # Módulo de Colab para subir/descargar archivos
import io                                 # Permite tratar los bytes del archivo subido como si fueran un archivo en memoria

print("Entorno listo")                    # Mensaje para confirmar que todas las librerías se importaron sin error

# Botón para cargar el Excel
print("Selecciona tu archivo Excel:")     # Aviso al usuario de que debe elegir el archivo
subido = files.upload()                    # Abre el botón/ventana de Colab para seleccionar el archivo y lo guarda en un diccionario

nombre_archivo = list(subido.keys())[0]   # Obtiene el nombre del archivo subido (primera llave del diccionario)
datos = pd.read_excel(io.BytesIO(subido[nombre_archivo])) # Convierte los bytes del archivo en un DataFrame de pandas

print(f"Archivo '{nombre_archivo}' cargado correctamente ") # Confirma que el archivo se leyó bien
print(datos.head(10))                       # Muestra las primeras 5 filas para revisar que los datos se ven correctos

X = datos[["horas_estudio"]]
y = datos["nota"]

modelo = LinearRegression()
modelo.fit(X, y)
print("Modelo entrenado listo")

pendiente = modelo.coef_[0]
base = modelo.intercept_

print(f"Por cada hora de estudio, la nota sube unos {pendiente:.1f} puntos")  # Interpreta la pendiente en palabras
print(f"Con 0 horas de estudio, el modelo espera una nota base de {base:.1f}")  # Interpreta el intercepto en palabras


horas_nuevas = pd.DataFrame({"horas_estudio":[6]})
predicion = modelo.predict(horas_nuevas)

print(f"Prediccion para 6 horas de estudio : {predicion[0]:.2f} puntos")
```
